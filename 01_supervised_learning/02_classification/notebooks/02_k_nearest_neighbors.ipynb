````python
# 📊 Load Social Network Ads Dataset for KNN Analysis
dataset = pd.read_csv('../01_Data/Social_Network_Ads.csv')

# 🎯 Feature Selection for KNN Distance Calculations
# Using Age and EstimatedSalary for 2D similarity analysis
X = dataset.iloc[:, [2, 3]].values  # Age (column 2) and EstimatedSalary (column 3)
y = dataset.iloc[:, -1].values      # Purchased (last column)

print("🎉 Social Network Ads dataset loaded for KNN analysis!")
print(f"📈 Dataset shape: {dataset.shape}")
print(f"👥 Customer records: {len(dataset):,}")
print(f"🎯 Features for similarity: Age and EstimatedSalary")
print(f"📊 Target variable: Purchased (0=No, 1=Yes)")

# Display sample of the data for KNN context
print(f"\n📋 Sample Customer Records (KNN Neighbors Pool):")
sample_data = dataset[['Age', 'EstimatedSalary', 'Purchased']].head(10)
print(sample_data.to_string(index=False))

# KNN-specific data analysis
print(f"\n🎯 KNN Dataset Analysis:")
print(f"• Age range: {X[:, 0].min():.0f} to {X[:, 0].max():.0f} years")
print(f"• Salary range: ${X[:, 1].min():,.0f} to ${X[:, 1].max():,.0f}")
print(f"• Purchase rate: {(y.sum() / len(y)) * 100:.1f}% of customers purchased")

# Critical scaling analysis for KNN
age_range = X[:, 0].max() - X[:, 0].min()
salary_range = X[:, 1].max() - X[:, 1].min()
scale_ratio = salary_range / age_range

print(f"\n⚠️ CRITICAL KNN Scale Analysis:")
print(f"• Age range: {age_range:.0f} units")
print(f"• Salary range: ${salary_range:,.0f}")
print(f"• Scale difference: {scale_ratio:,.0f}:1")
print(f"• Without scaling: Salary will dominate all distance calculations!")
print(f"• Impact: Age will become irrelevant in neighbor selection")
print(f"• Solution: Feature scaling is MANDATORY for KNN success")

# Preview distance calculation impact
print(f"\n🔍 Distance Calculation Preview (Unscaled):")
print(f"Customer A: Age 25, Salary $50K")
print(f"Customer B: Age 35, Salary $60K") 
print(f"Customer C: Age 30, Salary $80K")
print(f"Distance A→B: √[(35-25)² + (60000-50000)²] = √[100 + 100000000] ≈ 10,000")
print(f"Distance A→C: √[(30-25)² + (80000-50000)²] = √[25 + 900000000] ≈ 30,000")
print(f"💡 Notice: 10-year age difference = tiny impact vs $10K salary difference!")

print(f"\n🚀 Ready for feature scaling and KNN training!")
````

## Step 3: Train/Test Split

We split the dataset into 75% training (300 customers) and 25% test (100 customers).

**Why split before scaling?** The StandardScaler will be fit on the training set only. If we scaled before splitting, the test set's statistics would leak into the scaler — giving us an artificially optimistic evaluation. The split must come first to ensure the test set truly simulates unseen data.

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
````

## Step 4: Feature Scaling — Mandatory for KNN

**This is the most critical preprocessing step for KNN.** Here is the exact problem without scaling:

```
Customer A: Age 25, Salary $50,000
Customer B: Age 35, Salary $60,000
Customer C: Age 30, Salary $80,000

Unscaled distance A→B: sqrt[(35-25)² + (60000-50000)²] ≈ 10,000
Unscaled distance A→C: sqrt[(30-25)² + (80000-50000)²] ≈ 30,000
```

A 10-year age difference contributes almost nothing (100) compared to a $10,000 salary difference (100,000,000). Age is effectively invisible in all distance calculations.

After StandardScaler, both features are centred at 0 with standard deviation 1. Equal scales mean equal contribution to distances — the model uses both age and salary to find genuine neighbours.

**Rule:** Fit the scaler on training data only (`fit_transform`), apply it to test data with `transform`.

````python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
````

## Step 5: Train the KNN Model

```python
KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2)
```

**`n_neighbors=5`** — when classifying a new point, find the 5 most similar training examples and let them vote. The majority class wins.

**`metric='minkowski', p=2`** — this is **Euclidean distance** (the standard straight-line distance). Setting `p=1` gives Manhattan distance (sum of absolute differences). Minkowski with `p=2` is the default and works well for most cases.

**Why K=5?** It is a common default that balances:
- **Too small (K=1):** Model memorises training data. Every noisy sample gets its own region. High variance.
- **Too large (K=N):** Model always predicts the majority class regardless of input. High bias.
- **K=5:** Enough neighbours for a stable vote, small enough to capture local patterns.

In practice, you would choose K using cross-validation: try K=1, 3, 5, 7, 11 and pick the value that minimises validation error.

**Note:** KNN has no training phase — it simply stores all training data. Prediction is where the work happens.

````python
from sklearn.neighbors import KNeighborsClassifier
classifier = KNeighborsClassifier(n_neighbors = 5, metric = 'minkowski', p = 2)
classifier.fit(X_train, y_train)
````

## Step 6: Predict a Single Customer

We predict whether a 30-year-old earning $87,000 would purchase.

**Two critical requirements:**

1. **Apply the same scaler** — `sc.transform([[30, 87000]])` standardises the new point using the mean and std from the training set. Never refit the scaler on new data.
2. **2D array format** — `[[30, 87000]]` (double brackets) creates shape `(1, 2)`. A 1D array `[30, 87000]` would raise an error.

After scaling, KNN finds the 5 nearest training customers and returns their majority vote.

````python
print(classifier.predict(sc.transform([[30,87000]])))
````

## Step 7: Predict All Test Customers

We apply the model to all 100 test customers. For each one, KNN:
1. Scales the input using the training scaler
2. Computes Euclidean distance to all 300 training points
3. Finds the 5 nearest neighbours
4. Returns the majority class among those 5

The output shows `[predicted, actual]` pairs. Rows where they differ are misclassifications — which the confusion matrix will summarise.

````python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
````

## Step 8: Evaluate — Confusion Matrix

The confusion matrix gives a complete breakdown of correct and incorrect predictions:

```
              Predicted 0    Predicted 1
Actual 0        TN               FP
Actual 1        FN               TP
```

**93% accuracy** means KNN correctly classified 93 of 100 test customers using only age and salary as features.

**Checking for class imbalance:** If the dataset has far more 0s than 1s, 93% accuracy is not necessarily impressive — a model that always predicts 0 might also get ~70% accuracy. Look at the confusion matrix breakdown:
- High TN + high TP = genuinely useful model
- High TN + low TP = model ignores the minority class

**KNN vs Logistic Regression on this dataset:** Both achieve ~93% accuracy. KNN gets there through local similarity reasoning; Logistic Regression through a global linear decision boundary. Neither dominates — for a 2-feature problem with a moderately non-linear boundary, both work well.

````python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
````

## Step 9: Visualise the Training Set Decision Boundary

KNN creates a **non-linear, irregular decision boundary** — one of its most distinctive characteristics.

Unlike Logistic Regression (straight line) or SVM (smooth curve), KNN's boundary follows the local density of training points. Regions where one class dominates get that colour; mixed regions get a noisy, jagged boundary.

**What the visualisation shows:**
- **Red region:** KNN predicts 0 (will not purchase) for customers in this area of the age-salary space
- **Green region:** KNN predicts 1 (will purchase)
- **Dots:** Actual training labels (red = not purchased, green = purchased)

The training boundary looks very clean because the model was fit on this data. Test set is the honest check.

````python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_train), y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 1),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 1))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('K-NN (Training set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
````

## Step 10: Visualise the Test Set Decision Boundary

Test customers are plotted against the boundary learned from training data.

**Misclassified points** appear as dots of the wrong colour inside a region — e.g., a red dot (not purchased) inside a green region (KNN predicted purchased).

**Interpreting the KNN boundary shape:** The jagged, locally-adaptive boundary is both KNN's strength and weakness:
- **Strength:** Captures local patterns that global linear models miss
- **Weakness:** Sensitive to noise — a single outlier in the training set creates a local pocket in the boundary

Increasing K smooths the boundary; decreasing K makes it more jagged. The K=5 boundary shown here is a reasonable middle ground.

````python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_test), y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 1),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 1))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('K-NN (Test set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
````
