## 🎯 What Makes Naive Bayes Revolutionary?

### 🧮 **Built on Pure Mathematics**
Naive Bayes is founded on **Bayes' Theorem** (1763), making it one of the oldest yet most relevant ML algorithms:

```
P(Class|Features) = P(Features|Class) × P(Class) / P(Features)
```

### 🌟 **The "Naive" Assumption**
Despite being called "naive," this assumption is what makes the algorithm so powerful:
- **Feature Independence**: Assumes features don't influence each other
- **Computational Efficiency**: Reduces complex multivariate problems to simple univariate ones
- **Surprising Effectiveness**: Works amazingly well even when assumption is violated

## 🧮 Mathematical Foundation: Bayes' Theorem

**Understanding the Probabilistic Engine Behind Naive Bayes**

Before diving into implementation, let's build a solid mathematical foundation. Naive Bayes is fundamentally about **conditional probability** - predicting the likelihood of an outcome given observed evidence.

### 🎯 The Core Question
*"Given that a customer is 35 years old and earns $70,000, what's the probability they will purchase our product?"*

This is a **conditional probability** problem: P(Purchase | Age=35, Salary=70k)

### 📊 Bayes' Theorem - The Mathematical Engine

**Thomas Bayes' Revolutionary Formula:**

$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

**In our classification context:**

$$P(\text{Class}|\text{Features}) = \frac{P(\text{Features}|\text{Class}) \cdot P(\text{Class})}{P(\text{Features})}$$

### 🔍 Breaking Down Each Component

**1. 🎯 Posterior Probability: P(Class|Features)**
- What we want to find
- "Probability of purchase GIVEN age and salary"
- The final prediction

**2. 📈 Likelihood: P(Features|Class)**
- "How likely are these features for this class?"
- "What's the probability of Age=35, Salary=70k among customers who DID purchase?"

**3. 🎲 Prior Probability: P(Class)**
- Background knowledge before seeing features
- "Overall purchase rate in our data"
- Base rate fallacy protection

**4. 📊 Evidence: P(Features)**
- Normalizing constant
- "Probability of observing these features overall"
- Often ignored in classification (same for all classes)

### 🧠 The "Naive" Assumption

**Why "Naive"?** We assume features are **conditionally independent**:

$$P(\text{Age, Salary}|\text{Class}) = P(\text{Age}|\text{Class}) \times P(\text{Salary}|\text{Class})$$

**This means:**
- Age and Salary are independent GIVEN the class
- Knowing salary doesn't change the age distribution within a class
- Simplifies computation dramatically
- Often works well even when violated!

````
# Interactive Bayes' Theorem Demonstration
print("🧮 BAYES' THEOREM VISUAL DEMONSTRATION")
print("="*45)

# Let's demonstrate Bayes' theorem with our actual data
print("🎯 PRACTICAL BAYES' THEOREM EXAMPLE")
print("-"*35)

# Calculate actual probabilities from our data
# Prior probabilities
P_purchased = (y == 1).mean()
P_not_purchased = (y == 0).mean()

print(f"📊 PRIOR PROBABILITIES (Before seeing features):")
print(f"   P(Purchased) = {P_purchased:.3f} ({P_purchased*100:.1f}%)")
print(f"   P(Not Purchased) = {P_not_purchased:.3f} ({P_not_purchased*100:.1f}%)")

# Let's take a specific example: Age = 35, Salary = 70000
example_age = 35
example_salary = 70000

print(f"\n🔍 EXAMPLE PREDICTION:")
print(f"   Customer: Age = {example_age}, Salary = ${example_salary:,}")

# Calculate likelihood using Gaussian assumption
from scipy.stats import norm

# Parameters for each class
ages_no = X[y==0, 0]
ages_yes = X[y==1, 0]
salaries_no = X[y==0, 1]
salaries_yes = X[y==1, 1]

# Age likelihoods
age_likelihood_no = norm.pdf(example_age, ages_no.mean(), ages_no.std())
age_likelihood_yes = norm.pdf(example_age, ages_yes.mean(), ages_yes.std())

# Salary likelihoods
salary_likelihood_no = norm.pdf(example_salary, salaries_no.mean(), salaries_no.std())
salary_likelihood_yes = norm.pdf(example_salary, salaries_yes.mean(), salaries_yes.std())

print(f"\n📈 LIKELIHOOD CALCULATIONS:")
print(f"🔢 Age = {example_age} years:")
print(f"   P(Age={example_age} | Not Purchased) = {age_likelihood_no:.6f}")
print(f"   P(Age={example_age} | Purchased) = {age_likelihood_yes:.6f}")

print(f"💰 Salary = ${example_salary:,}:")
print(f"   P(Salary=${example_salary:,} | Not Purchased) = {salary_likelihood_no:.2e}")
print(f"   P(Salary=${example_salary:,} | Purchased) = {salary_likelihood_yes:.2e}")

# Combined likelihood (naive independence assumption)
likelihood_no = age_likelihood_no * salary_likelihood_no
likelihood_yes = age_likelihood_yes * salary_likelihood_yes

print(f"\n🧠 COMBINED LIKELIHOODS (Naive Independence):")
print(f"   P(Age={example_age}, Salary=${example_salary:,} | Not Purchased) = {likelihood_no:.2e}")
print(f"   P(Age={example_age}, Salary=${example_salary:,} | Purchased) = {likelihood_yes:.2e}")

# Posterior probabilities (Bayes' theorem)
posterior_no = likelihood_no * P_not_purchased
posterior_yes = likelihood_yes * P_purchased

# Normalize (evidence calculation)
evidence = posterior_no + posterior_yes
posterior_no_normalized = posterior_no / evidence
posterior_yes_normalized = posterior_yes / evidence

print(f"\n🎯 BAYES' THEOREM APPLICATION:")
print(f"   Numerator (Not Purchased): {likelihood_no:.2e} × {P_not_purchased:.3f} = {posterior_no:.2e}")
print(f"   Numerator (Purchased): {likelihood_yes:.2e} × {P_purchased:.3f} = {posterior_yes:.2e}")
print(f"   Evidence (Normalizer): {evidence:.2e}")

print(f"\n🏆 FINAL POSTERIOR PROBABILITIES:")
print(f"   P(Not Purchased | Age={example_age}, Salary=${example_salary:,}) = {posterior_no_normalized:.3f} ({posterior_no_normalized*100:.1f}%)")
print(f"   P(Purchased | Age={example_age}, Salary=${example_salary:,}) = {posterior_yes_normalized:.3f} ({posterior_yes_normalized*100:.1f}%)")

prediction = "PURCHASED" if posterior_yes_normalized > 0.5 else "NOT PURCHASED"
confidence = max(posterior_no_normalized, posterior_yes_normalized)
print(f"\n🎯 PREDICTION: {prediction} (Confidence: {confidence*100:.1f}%)")

# Visualize the probability distributions
plt.figure(figsize=(15, 10))

# Age distributions
plt.subplot(2, 3, 1)
age_range = np.linspace(18, 60, 200)
plt.plot(age_range, norm.pdf(age_range, ages_no.mean(), ages_no.std()), 
         'r-', linewidth=2, label=f'Not Purchased (μ={ages_no.mean():.1f})')
plt.plot(age_range, norm.pdf(age_range, ages_yes.mean(), ages_yes.std()), 
         'g-', linewidth=2, label=f'Purchased (μ={ages_yes.mean():.1f})')
plt.axvline(example_age, color='blue', linestyle='--', linewidth=2, label=f'Example Age = {example_age}')
plt.xlabel('Age')
plt.ylabel('Probability Density')
plt.title('Age Likelihood Distributions')
plt.legend()
plt.grid(True, alpha=0.3)

# Salary distributions
plt.subplot(2, 3, 2)
salary_range = np.linspace(15000, 150000, 200)
plt.plot(salary_range, norm.pdf(salary_range, salaries_no.mean(), salaries_no.std()), 
         'r-', linewidth=2, label=f'Not Purchased (μ=${salaries_no.mean():.0f})')
plt.plot(salary_range, norm.pdf(salary_range, salaries_yes.mean(), salaries_yes.std()), 
         'g-', linewidth=2, label=f'Purchased (μ=${salaries_yes.mean():.0f})')
plt.axvline(example_salary, color='blue', linestyle='--', linewidth=2, label=f'Example Salary = ${example_salary:,}')
plt.xlabel('Salary')
plt.ylabel('Probability Density')
plt.title('Salary Likelihood Distributions')
plt.legend()
plt.grid(True, alpha=0.3)

# Prior probabilities
plt.subplot(2, 3, 3)
categories = ['Not Purchased', 'Purchased']
priors = [P_not_purchased, P_purchased]
colors = ['red', 'green']
bars = plt.bar(categories, priors, color=colors, alpha=0.7)
plt.ylabel('Prior Probability')
plt.title('Prior Probabilities\n(Base Rates)')
plt.grid(True, alpha=0.3)

# Add values on bars
for bar, prior in zip(bars, priors):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.01,
             f'{prior:.3f}', ha='center', va='bottom')

# Likelihood comparison for our example
plt.subplot(2, 3, 4)
likelihood_values = [likelihood_no, likelihood_yes]
bars = plt.bar(categories, likelihood_values, color=colors, alpha=0.7)
plt.ylabel('Likelihood')
plt.title(f'Likelihoods for Example\n(Age={example_age}, Salary=${example_salary:,})')
plt.yscale('log')  # Log scale due to very small values
plt.grid(True, alpha=0.3)

# Add values on bars
for bar, likelihood in zip(bars, likelihood_values):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height * 1.5,
             f'{likelihood:.2e}', ha='center', va='bottom', rotation=0)

# Posterior probabilities
plt.subplot(2, 3, 5)
posterior_values = [posterior_no_normalized, posterior_yes_normalized]
bars = plt.bar(categories, posterior_values, color=colors, alpha=0.7)
plt.ylabel('Posterior Probability')
plt.title(f'Final Predictions\n(After Bayes Theorem)')
plt.grid(True, alpha=0.3)

# Add values on bars
for bar, posterior in zip(bars, posterior_values):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.02,
             f'{posterior:.3f}', ha='center', va='bottom')

# Bayes theorem flow diagram
plt.subplot(2, 3, 6)
plt.text(0.5, 0.9, "Bayes' Theorem Flow", ha='center', va='center', 
         fontsize=14, fontweight='bold', transform=plt.gca().transAxes)

flow_text = f"""
Prior × Likelihood = Posterior

P(Purchased) = {P_purchased:.3f}
×
P(Features|Purchased) = {likelihood_yes:.2e}
=
P(Purchased|Features) = {posterior_yes_normalized:.3f}

vs.

P(Not Purchased) = {P_not_purchased:.3f}
×
P(Features|Not Purchased) = {likelihood_no:.2e}
=
P(Not Purchased|Features) = {posterior_no_normalized:.3f}
"""

plt.text(0.05, 0.05, flow_text, ha='left', va='bottom', 
         fontsize=9, fontfamily='monospace',
         transform=plt.gca().transAxes,
         bbox=dict(boxstyle="round,pad=0.3", facecolor='lightblue', alpha=0.5))

plt.axis('off')

plt.tight_layout()
plt.show()

print(f"\n🧠 KEY INSIGHTS:")
print(f"   📊 Prior knowledge: {P_purchased*100:.1f}% purchase rate")
print(f"   📈 Likelihood impact: {'Purchase' if likelihood_yes > likelihood_no else 'No Purchase'} features more likely")
print(f"   🎯 Posterior update: Bayes theorem combines both for final prediction")
print(f"   🔍 Independence assumption: Age and salary treated as independent given class")

print(f"\n🎯 Ready for Gaussian Naive Bayes implementation!")
````

## 📊 Gaussian Naive Bayes: The Mathematical Deep Dive

**Understanding How Continuous Features Work with Naive Bayes**

Now that we understand Bayes' theorem, let's explore how **Gaussian Naive Bayes** specifically handles continuous features like age and salary.

### 🎲 The Gaussian Assumption

For continuous features, we assume each feature follows a **normal (Gaussian) distribution** within each class:

$$P(\text{feature}|\text{class}) \sim \mathcal{N}(\mu_{\text{class}}, \sigma_{\text{class}}^2)$$

**The Gaussian Probability Density Function:**

$$P(x|\text{class}) = \frac{1}{\sqrt{2\pi\sigma_{\text{class}}^2}} \exp\left(-\frac{(x - \mu_{\text{class}})^2}{2\sigma_{\text{class}}^2}\right)$$

### 🔍 Parameter Estimation

**For each feature and each class, we estimate:**

1. **Mean (μ)**: Average value of the feature for that class
2. **Standard Deviation (σ)**: Spread of the feature for that class

**Training Process:**
- Calculate μ and σ for each feature-class combination
- Store these parameters for prediction

### 🧠 The Complete Gaussian NB Formula

**For a sample with features x₁, x₂, ..., xₙ:**

$$P(\text{class}|\mathbf{x}) = \frac{P(\text{class}) \prod_{i=1}^{n} P(x_i|\text{class})}{P(\mathbf{x})}$$

**Where each P(xᵢ|class) is calculated using the Gaussian PDF**

### 🎯 Key Advantages of Gaussian NB

**1. 📊 Probabilistic Output**
- Not just predictions, but confidence levels
- Natural uncertainty quantification

**2. ⚡ Computational Efficiency**
- O(n) training time
- O(1) prediction time per sample

**3. 🎲 Handles Small Datasets**
- Simple parameter estimation
- Less prone to overfitting

**4. 📈 Interpretable Parameters**
- Clear meaning: mean and variance per class
- Easy to explain to stakeholders

### ⚠️ Key Assumptions & Limitations

**1. 🧠 Naive Independence**
- Features must be conditionally independent
- Often violated but still works well

**2. 📊 Gaussian Distribution**
- Features should be roughly normal within classes
- Can be violated with decent results

**3. 🎯 Continuous Features**
- Designed for continuous data
- Categorical features need different variants

````
# Gaussian Naive Bayes Parameter Analysis
print("📊 GAUSSIAN NAIVE BAYES PARAMETER DEEP DIVE")
print("="*50)

# Let's manually calculate what Gaussian NB will learn
print("🧠 MANUAL PARAMETER CALCULATION")
print("-"*35)

# Separate data by class
X_no_purchase = X[y == 0]  # Features for class 0
X_purchased = X[y == 1]    # Features for class 1

print(f"📈 DATASET SPLIT BY CLASS:")
print(f"   No Purchase samples: {len(X_no_purchase)}")
print(f"   Purchased samples: {len(X_purchased)}")

# Calculate parameters for each feature and class
print(f"\n🎯 GAUSSIAN PARAMETERS BY CLASS:")

# Age parameters
age_mean_no = X_no_purchase[:, 0].mean()
age_std_no = X_no_purchase[:, 0].std()
age_mean_yes = X_purchased[:, 0].mean()
age_std_yes = X_purchased[:, 0].std()

print(f"🔢 AGE PARAMETERS:")
print(f"   No Purchase:  μ = {age_mean_no:.2f}, σ = {age_std_no:.2f}")
print(f"   Purchased:    μ = {age_mean_yes:.2f}, σ = {age_std_yes:.2f}")

# Salary parameters
salary_mean_no = X_no_purchase[:, 1].mean()
salary_std_no = X_no_purchase[:, 1].std()
salary_mean_yes = X_purchased[:, 1].mean()
salary_std_yes = X_purchased[:, 1].std()

print(f"💰 SALARY PARAMETERS:")
print(f"   No Purchase:  μ = ${salary_mean_no:.0f}, σ = ${salary_std_no:.0f}")
print(f"   Purchased:    μ = ${salary_mean_yes:.0f}, σ = ${salary_std_yes:.0f}")

# Class priors
prior_no = len(X_no_purchase) / len(X)
prior_yes = len(X_purchased) / len(X)

print(f"\n🎲 CLASS PRIORS:")
print(f"   P(No Purchase) = {prior_no:.3f}")
print(f"   P(Purchased) = {prior_yes:.3f}")

# Visualize the Gaussian distributions learned
plt.figure(figsize=(16, 12))

# Age distributions
plt.subplot(2, 3, 1)
age_range = np.linspace(X[:, 0].min() - 5, X[:, 0].max() + 5, 200)

# Plot Gaussian curves
gaussian_no_age = norm.pdf(age_range, age_mean_no, age_std_no)
gaussian_yes_age = norm.pdf(age_range, age_mean_yes, age_std_yes)

plt.plot(age_range, gaussian_no_age, 'r-', linewidth=3, 
         label=f'No Purchase\nμ={age_mean_no:.1f}, σ={age_std_no:.1f}')
plt.plot(age_range, gaussian_yes_age, 'g-', linewidth=3, 
         label=f'Purchased\nμ={age_mean_yes:.1f}, σ={age_std_yes:.1f}')

# Add actual data points
plt.hist(X_no_purchase[:, 0], bins=15, alpha=0.3, color='red', density=True, label='Actual (No)')
plt.hist(X_purchased[:, 0], bins=15, alpha=0.3, color='green', density=True, label='Actual (Yes)')

plt.xlabel('Age')
plt.ylabel('Probability Density')
plt.title('Age: Learned Gaussian Distributions')
plt.legend()
plt.grid(True, alpha=0.3)

# Salary distributions
plt.subplot(2, 3, 2)
salary_range = np.linspace(X[:, 1].min() - 5000, X[:, 1].max() + 5000, 200)

# Plot Gaussian curves
gaussian_no_salary = norm.pdf(salary_range, salary_mean_no, salary_std_no)
gaussian_yes_salary = norm.pdf(salary_range, salary_mean_yes, salary_std_yes)

plt.plot(salary_range, gaussian_no_salary, 'r-', linewidth=3, 
         label=f'No Purchase\nμ=${salary_mean_no:.0f}, σ=${salary_std_no:.0f}')
plt.plot(salary_range, gaussian_yes_salary, 'g-', linewidth=3, 
         label=f'Purchased\nμ=${salary_mean_yes:.0f}, σ=${salary_std_yes:.0f}')

# Add actual data points
plt.hist(X_no_purchase[:, 1], bins=15, alpha=0.3, color='red', density=True, label='Actual (No)')
plt.hist(X_purchased[:, 1], bins=15, alpha=0.3, color='green', density=True, label='Actual (Yes)')

plt.xlabel('Salary')
plt.ylabel('Probability Density')
plt.title('Salary: Learned Gaussian Distributions')
plt.legend()
plt.grid(True, alpha=0.3)

# 2D feature space with Gaussian contours
plt.subplot(2, 3, 3)

# Create a grid for contour plotting
age_grid = np.linspace(X[:, 0].min(), X[:, 0].max(), 50)
salary_grid = np.linspace(X[:, 1].min(), X[:, 1].max(), 50)
Age_grid, Salary_grid = np.meshgrid(age_grid, salary_grid)

# Calculate 2D Gaussian probabilities (assuming independence)
from scipy.stats import multivariate_normal

# Covariance matrices (diagonal due to independence assumption)
cov_no = np.array([[age_std_no**2, 0], [0, salary_std_no**2]])
cov_yes = np.array([[age_std_yes**2, 0], [0, salary_std_yes**2]])

# Create 2D probability surfaces
pos = np.dstack((Age_grid, Salary_grid))
rv_no = multivariate_normal([age_mean_no, salary_mean_no], cov_no)
rv_yes = multivariate_normal([age_mean_yes, salary_mean_yes], cov_yes)

# Plot contours
plt.contour(Age_grid, Salary_grid, rv_no.pdf(pos), colors='red', alpha=0.6, levels=5)
plt.contour(Age_grid, Salary_grid, rv_yes.pdf(pos), colors='green', alpha=0.6, levels=5)

# Plot actual data points
plt.scatter(X_no_purchase[:, 0], X_no_purchase[:, 1], c='red', alpha=0.6, s=30, label='No Purchase')
plt.scatter(X_purchased[:, 0], X_purchased[:, 1], c='green', alpha=0.6, s=30, label='Purchased')

plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('2D Feature Space\nGaussian Contours (Independence)')
plt.legend()
plt.grid(True, alpha=0.3)

# Parameter comparison
plt.subplot(2, 3, 4)
features = ['Age', 'Salary']
means_no = [age_mean_no, salary_mean_no/1000]  # Scale salary for comparison
means_yes = [age_mean_yes, salary_mean_yes/1000]
stds_no = [age_std_no, salary_std_no/1000]
stds_yes = [age_std_yes, salary_std_yes/1000]

x_pos = np.arange(len(features))
width = 0.35

bars1 = plt.bar(x_pos - width/2, means_no, width, yerr=stds_no, 
                capsize=5, label='No Purchase', color='red', alpha=0.7)
bars2 = plt.bar(x_pos + width/2, means_yes, width, yerr=stds_yes, 
                capsize=5, label='Purchased', color='green', alpha=0.7)

plt.xlabel('Features')
plt.ylabel('Mean ± Std (Salary in $1000s)')
plt.title('Gaussian Parameters Comparison')
plt.xticks(x_pos, features)
plt.legend()
plt.grid(True, alpha=0.3)

# Likelihood visualization for different feature values
plt.subplot(2, 3, 5)
test_ages = np.linspace(20, 60, 100)
likelihood_ratios = []

for age in test_ages:
    # Calculate likelihood ratio: P(age|purchased) / P(age|not_purchased)
    like_yes = norm.pdf(age, age_mean_yes, age_std_yes)
    like_no = norm.pdf(age, age_mean_no, age_std_no)
    ratio = like_yes / (like_no + 1e-10)  # Small epsilon to avoid division by zero
    likelihood_ratios.append(ratio)

plt.plot(test_ages, likelihood_ratios, 'b-', linewidth=2)
plt.axhline(y=1, color='black', linestyle='--', alpha=0.5, label='Equal likelihood')
plt.axhline(y=prior_no/prior_yes, color='orange', linestyle='--', alpha=0.7, 
            label=f'Prior ratio = {prior_no/prior_yes:.2f}')
plt.xlabel('Age')
plt.ylabel('Likelihood Ratio\nP(age|purchased) / P(age|not_purchased)')
plt.title('Age Likelihood Ratios')
plt.legend()
plt.grid(True, alpha=0.3)

# Fill areas where each class is favored
plt.fill_between(test_ages, 0, likelihood_ratios, 
                 where=np.array(likelihood_ratios) > 1, 
                 alpha=0.2, color='green', label='Favors Purchase')
plt.fill_between(test_ages, 0, likelihood_ratios, 
                 where=np.array(likelihood_ratios) < 1, 
                 alpha=0.2, color='red', label='Favors No Purchase')

# Decision boundary analysis
plt.subplot(2, 3, 6)
plt.text(0.1, 0.9, "Gaussian NB Decision Process", fontsize=14, fontweight='bold', 
         transform=plt.gca().transAxes)

decision_text = f"""
1. TRAINING PHASE:
   • Calculate μ, σ for each feature-class pair
   • Store prior probabilities

2. PREDICTION PHASE:
   For new sample (age, salary):
   
   • P(age|class) using Gaussian PDF
   • P(salary|class) using Gaussian PDF
   • P(age,salary|class) = P(age|class) × P(salary|class)
   • Posterior = Prior × Likelihood
   • Predict class with highest posterior

3. LEARNED PARAMETERS:
   Age (No):    μ={age_mean_no:.1f}, σ={age_std_no:.1f}
   Age (Yes):   μ={age_mean_yes:.1f}, σ={age_std_yes:.1f}
   Salary (No): μ=${salary_mean_no:.0f}, σ=${salary_std_no:.0f}
   Salary (Yes): μ=${salary_mean_yes:.0f}, σ=${salary_std_yes:.0f}
"""

plt.text(0.05, 0.05, decision_text, fontsize=9, fontfamily='monospace',
         transform=plt.gca().transAxes, verticalalignment='bottom',
         bbox=dict(boxstyle="round,pad=0.5", facecolor='lightblue', alpha=0.7))

plt.axis('off')

plt.tight_layout()
plt.show()

print(f"\n🧠 GAUSSIAN NB INSIGHTS:")
print(f"   📊 Feature separability:")
print(f"      Age difference: {abs(age_mean_yes - age_mean_no):.1f} years")
print(f"      Salary difference: ${abs(salary_mean_yes - salary_mean_no):.0f}")
print(f"   📈 Overlap assessment:")
print(f"      Age overlap: {'High' if abs(age_mean_yes - age_mean_no) < (age_std_no + age_std_yes)/2 else 'Low'}")
print(f"      Salary overlap: {'High' if abs(salary_mean_yes - salary_mean_no) < (salary_std_no + salary_std_yes)/2 else 'Low'}")
print(f"   🎯 Expected performance: {'Good' if abs(age_mean_yes - age_mean_no) > 2 else 'Moderate'} separation")

print(f"\n🎯 Ready for scikit-learn implementation!")
````

## Step 6: Train Gaussian Naive Bayes

Naive Bayes computes two things during training:
1. **P(class)** — the prior probability of each class (fraction of training samples per class)
2. **P(feature | class)** — for Gaussian NB, the mean and variance of each feature for each class

There are no hyperparameters to tune and no gradient descent — training is a single pass through the data to compute these statistics.

During prediction, Bayes' theorem combines these:

$$P(\text{class} | \text{features}) \propto P(\text{class}) \times \prod_i P(x_i | \text{class})$$

The class with the highest posterior probability wins.

````
from sklearn.naive_bayes import GaussianNB
classifier = GaussianNB()
classifier.fit(X_train, y_train)
````

````
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_train), y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 0.25),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 0.25))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Naive Bayes (Training set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
````
